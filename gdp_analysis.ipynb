## GDP Development Across Countries (2000–2022)

The GDP data from the World Bank reveal substantial differences in economic growth across the seven countries between 2000 and 2022.

- **United States:** Maintained the largest nominal GDP throughout the period, experiencing sustained long-term growth despite contractions during the 2008–09 Global Financial Crisis and the COVID-19 pandemic in 2020.

- **China:** Experienced the most dramatic GDP expansion, driven by rapid industrialisation, urbanisation and international trade. China significantly narrowed the economic gap with the United States, although its growth moderated in later years.

- **Japan:** Exhibited relatively stagnant nominal GDP in US dollar terms, with substantial fluctuations reflecting exchange-rate movements and prolonged periods of weak economic growth.

- **Germany:** Demonstrated generally positive long-term GDP growth, with notable setbacks during the Global Financial Crisis and COVID-19 pandemic. Its export-oriented economy was particularly exposed to global economic conditions.

- **United Kingdom:** Experienced moderate but uneven GDP growth, interrupted by the Global Financial Crisis, Brexit-related uncertainty and the COVID-19 pandemic. Changes in the pound's exchange rate also affected GDP measured in US dollars.

- **Brazil:** Recorded strong GDP growth during the 2000s, supported partly by the commodity boom, followed by considerable volatility and economic contractions during the mid-2010s.

- **Switzerland:** Experienced relatively steady long-term GDP expansion, although its GDP in US dollar terms was influenced by fluctuations in the Swiss franc.

### Key Takeaway

While the United States maintained its position as the largest economy, China exhibited the strongest expansion. Advanced economies generally experienced more moderate growth, while Brazil demonstrated greater volatility.

**Limitation:** The analysis uses nominal GDP expressed in current US dollars. Consequently, changes reflect not only real economic growth but also inflation and exchange-rate fluctuations. Real GDP or GDP indexed to a base year would provide a more appropriate basis for comparing economic growth rates across countries.
